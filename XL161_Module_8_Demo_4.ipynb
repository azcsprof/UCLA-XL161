```python
# ============================================================
# MODULE 8 DEMO 4 — HIGH CONFIDENCE CAN STILL FAIL
# ============================================================

training_data = [
    "common_case",
    "common_case",
    "common_case",
    "common_case"
]

rare_case_count = 1

print("Training examples:")
print(training_data)

print()

print("Rare case count:")
print(rare_case_count)

print()

# The system mostly learns
# from common patterns.

if rare_case_count < 2:

    prediction = "safe"

    confidence = 0.97

else:

    prediction = "warning"

    confidence = 0.60

print("Prediction:")
print(prediction)

print()

print("Confidence:")
print(confidence)

print()

actual_outcome = "critical_condition"

print("Actual outcome:")
print(actual_outcome)

print()

if prediction == actual_outcome:

    print("Prediction succeeded.")

else:

    print("Prediction failed.")

print()

# High confidence does not
# guarantee correctness.

# Missing or underrepresented
# cases may distort the learned
# probabilities internally.
```
