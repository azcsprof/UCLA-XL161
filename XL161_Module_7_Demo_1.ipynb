```python
# ==================================================
# Module 7 DEMO 1: When Facts Don't Connect
# ==================================================
#
# Earlier logic systems could store facts
# and check whether specific statements were true.
#
# But storing facts is not the same thing
# as representing relationships in a usable way.
#
# In this demo, the system stores relationship
# information as flat strings.
#
# The system can answer exact lookup questions like:
# "Does Alice love Bob?"
#
# But it cannot answer general relationship questions like:
# "Who does Alice love?"
#
# Your job:
# 1. Notice what the system can check directly
# 2. Notice what it cannot ask flexibly
# 3. Explain why adding more facts does not fix the problem
# 4. Connect this failure to representation, not logic
#
# Important:
# The information is present.
# The structure needed to reason across it is missing.
# ==================================================


print("\n=== MODULE 7 DEMO 1 ===")
print("When Facts Don't Connect")


# --------------------------------------------------
# STEP 1:
# Store facts as flat strings
# --------------------------------------------------
#
# Each fact is stored as one complete label.
#
# The system does not know that:
# • Alice is the subject
# • loves is the relationship
# • Bob and Charlie are targets
#
# To the system, each fact is just a key.
# --------------------------------------------------

facts = {
    "Loves(Alice,Bob)": True,
    "Loves(Alice,Charlie)": True
}


print("\nStored facts:")

for fact in facts:
    print(fact)


# --------------------------------------------------
# STEP 2:
# Exact lookup works
# --------------------------------------------------
#
# The system can check a complete fact
# if we ask for it exactly as stored.
# --------------------------------------------------

print("\nExact lookup:")
print("Does Alice love Bob?")

result = facts.get("Loves(Alice,Bob)")

print(result)


# --------------------------------------------------
# STEP 3:
# General question fails
# --------------------------------------------------
#
# Now ask a more flexible question:
#
# Who does Alice love?
#
# This question requires the system to search
# across multiple facts and collect the targets.
#
# But the representation does not separate
# the parts of the relationship.
# --------------------------------------------------

query = "WhoDoesAliceLove"

print("\nGeneral query:")
print(query)

answer = facts.get(query)

print("Answer:", answer)


# --------------------------------------------------
# STEP 4:
# Add more data
# --------------------------------------------------
#
# Adding another flat fact increases the amount
# of information in the system.
#
# But it still does not create structure.
# --------------------------------------------------

facts["Loves(Alice,Dana)"] = True

print("\nAfter adding another fact:")

for fact in facts:
    print(fact)


print("\nTry the same general query again:")
print(query)

answer = facts.get(query)

print("Answer:", answer)


# --------------------------------------------------
# Think about this:
# --------------------------------------------------
#
# • The system has several facts about Alice
# • A human can see the pattern immediately
# • The system cannot separate subject, relation, and target
# • More facts do not automatically create better reasoning
#
# The problem is not that the system lacks information.
#
# The problem is that the information is stored
# in a form the system cannot reason across.
#
# Key question:
# What would need to change so the system could answer
# "Who does Alice love?"
# --------------------------------------------------


print("\n=== DEMO COMPLETE ===")
```
