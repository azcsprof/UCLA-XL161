```python
# ==================================================
# Module 7 DEMO 2: From Names to Relationships
# ==================================================
#
# Demo 1 showed that flat facts are limited.
#
# The system had information, but could not
# organize that information into useful answers.
#
# In this demo, facts are represented as structured tuples.
#
# Instead of storing:
# "Loves(Alice,Bob)"
#
# the system stores:
# ("Loves", "alice", "bob")
#
# This changes the representation.
#
# Now the system can identify:
# • the relationship
# • the subject
# • the target
#
# Your job:
# 1. Watch how structure changes what the system can ask
# 2. Notice why position matters
# 3. Explain why this representation is more flexible
# 4. Connect structured facts to relational reasoning
#
# Important:
# The system is not just storing more information.
# It is storing information in a more useful form.
# ==================================================


print("\n=== MODULE 7 DEMO 2 ===")
print("From Names to Relationships")


# --------------------------------------------------
# STEP 1:
# Store facts as structured relationships
# --------------------------------------------------
#
# Each fact now has parts.
#
# Position 0: relationship type
# Position 1: subject
# Position 2: target
# --------------------------------------------------

facts = [
    ("Loves", "alice", "bob"),
    ("Loves", "alice", "charlie"),
    ("Loves", "bob", "dana")
]


print("\nStructured facts:")

for fact in facts:
    print(fact)


# --------------------------------------------------
# STEP 2:
# Try the old kind of lookup
# --------------------------------------------------
#
# The old query from Demo 1 still does not exist
# as a stored fact.
#
# But that is no longer the point.
#
# Now the system has structure it can search through.
# --------------------------------------------------

old_query = "WhoDoesAliceLove"

print("\nOld-style lookup:")
print(old_query)

print(old_query in facts)


# --------------------------------------------------
# STEP 3:
# Search by relationship structure
# --------------------------------------------------
#
# Now the system can ask:
#
# Find every fact where:
# • the relationship is Loves
# • the subject is alice
#
# Then return the target.
# --------------------------------------------------

print("\nStructured query:")
print("Who does alice love?")

for relation, subject, target in facts:

    if relation == "Loves" and subject == "alice":
        print(target)


# --------------------------------------------------
# STEP 4:
# Add another relationship
# --------------------------------------------------
#
# We add a new fact with the same structure.
#
# The query code does not need to change.
# It already knows how to search by position.
# --------------------------------------------------

facts.append(("Loves", "alice", "emma"))

print("\nAfter adding another relationship:")
print("Who does alice love now?")

for relation, subject, target in facts:

    if relation == "Loves" and subject == "alice":
        print(target)


# --------------------------------------------------
# STEP 5:
# Reverse the relationship
# --------------------------------------------------
#
# Order matters.
#
# Loves(alice, bob)
# is not the same as
# Loves(bob, alice)
#
# The positions define the meaning.
# --------------------------------------------------

facts.append(("Loves", "bob", "alice"))

print("\nAfter adding reversed relationship:")
print("All facts:")

for fact in facts:
    print(fact)


print("\nWho does alice love?")

for relation, subject, target in facts:

    if relation == "Loves" and subject == "alice":
        print(target)


# --------------------------------------------------
# Think about this:
# --------------------------------------------------
#
# • The same names can appear in different positions
# • Position changes meaning
# • Relationships are now searchable
# • The system can answer a flexible question
#
# This is the beginning of structured representation.
#
# The system is no longer limited to exact lookup.
#
# It can search across the internal parts of a fact.
#
# Key question:
# Why does ("Loves", "alice", "bob") allow more reasoning
# than "Alice_loves_Bob"?
# --------------------------------------------------


print("\n=== DEMO COMPLETE ===")
```
