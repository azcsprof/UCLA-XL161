```python
from itertools import product

# -------------------------------------------
# Module 6 Demo 3 - From Checking to Inference
# -------------------------------------------

# -------------------------------------------
# PART A — Checking (Truth Table)
# -------------------------------------------
# The system evaluates EVERY possible case.
# This is exhaustive checking (brute force).

print("PART A — Checking (Truth Table):\n")

for A, B in product([True, False], repeat=2):
    print(f"A={A}, B={B} -> {A or B}")

# Key idea:
# The system is not reasoning — it is checking all possibilities.


# -------------------------------------------
# PART B — Deriving (Inference)
# -------------------------------------------
# Instead of checking everything, the system
# derives new facts from what is already true.

print("\nPART B — Deriving (Inference):\n")

facts = {"fever": True}

rules = [
    ("fever", "infection"),
    ("infection", "treatment")
]

changed = True
while changed:
    changed = False
    for condition, result in rules:
        if facts.get(condition) and result not in facts:
            facts[result] = True
            changed = True

print("Derived facts:", facts)

# Key idea:
# The system is generating consequences, not exploring options.


# -------------------------------------------
# PART C — Interaction (Emergence)
# -------------------------------------------
# Multiple facts combine to produce a new conclusion.

print("\nPART C — Interaction:\n")

facts = {"cough": True, "fever": True}

# Base facts
if facts["cough"]:
    facts["respiratory_issue"] = True

if facts["fever"]:
    facts["infection"] = True

# Combined condition
if facts.get("respiratory_issue") and facts.get("infection"):
    facts["serious_condition"] = True

print("Final facts:", facts)

# Key idea:
# The system can derive something not explicitly given.
# The conclusion comes from how rules interact.
```
