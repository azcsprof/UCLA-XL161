```python
# --------------------------------------------------
# Module 9 DEMO 3: Memorization vs Generalization
# --------------------------------------------------
# A model can perform perfectly on known examples
# while still failing on new situations.
#
# Your job:
# 1. Compare familiar vs unfamiliar cases
# 2. Watch where the system succeeds
# 3. Decide whether the model learned
#    a reusable pattern or only memorized data
# --------------------------------------------------


def diagnose(patient_id):

    # Memorized training case
    if patient_id == 442:
        return "flu"

    return "unknown"


# Familiar example
print("Known patient:")
print("Diagnosis:", diagnose(442))


# New example
print("\nNew patient:")
print("Diagnosis:", diagnose(901))


# --------------------------------------------------
# Now compare with a simpler generalized rule
# --------------------------------------------------

def generalized(fever):

    if fever:
        return "flu"

    return "unknown"


print("\nGeneralized system:")
print("Patient with fever:", generalized(True))
print("Patient without fever:", generalized(False))


# --------------------------------------------------
# Think about this:
# --------------------------------------------------
# • The first system memorized one exact case
#
# • The second system applies a broader pattern
#
# • Memorization may produce perfect
#   training performance
#
# • But real intelligence requires handling
#   unfamiliar situations
#
# Key question:
# Is the system learning reusable structure...
# or storing isolated examples?
# --------------------------------------------------
```
