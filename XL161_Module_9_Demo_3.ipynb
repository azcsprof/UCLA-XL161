```python
# ==================================================
# Module 9 DEMO 3: Memorization vs Generalization
# ==================================================
#
# A learning system is not successful simply
# because it performs well on familiar examples.
#
# The real challenge is generalization:
#
# Can the system handle situations it has
# never encountered before?
#
# In this demo, you will compare:
# • a system that memorizes exact cases
# • a system that applies broader patterns
#
# Your job:
# 1. Compare familiar vs unfamiliar examples
# 2. Watch where the system succeeds
# 3. Decide whether the model learned
#    reusable structure or only memorized data
#
# Important:
# Perfect performance on training examples
# does NOT guarantee robust intelligence.
# ==================================================


print("\n=== MODULE 9 DEMO 3 ===")
print("Memorization vs Generalization")


# --------------------------------------------------
# STEP 1:
# Create a memorization-based system
# --------------------------------------------------
#
# This model stores one exact training case.
#
# The system succeeds ONLY when the input
# exactly matches the memorized example.
#
# This often happens when models overfit
# training data instead of learning broader
# structure.
# --------------------------------------------------

def memorize(patient_id):

    # ----------------------------------------------
    # Memorized training example
    # ----------------------------------------------

    if patient_id == 442:

        return "flu"


    # ----------------------------------------------
    # Unknown cases fail
    # ----------------------------------------------

    return "unknown"


# --------------------------------------------------
# STEP 2:
# Test the memorized system
# --------------------------------------------------
#
# First:
# test the familiar example
#
# Then:
# test a new unseen case
# --------------------------------------------------

print("\n=== MEMORIZED SYSTEM ===")


# Familiar case
print("\nKnown patient:")

known_result = memorize(442)

print("Patient ID: 442")
print("Diagnosis:", known_result)


# New case
print("\nNew patient:")

new_result = memorize(901)

print("Patient ID: 901")
print("Diagnosis:", new_result)


# --------------------------------------------------
# STEP 3:
# Create a generalized system
# --------------------------------------------------
#
# Instead of memorizing one patient,
# this system applies a broader pattern.
#
# The rule is simpler, but it can now
# operate across many different cases.
# --------------------------------------------------

def generalize(fever):

    # ----------------------------------------------
    # Generalized pattern
    # ----------------------------------------------

    if fever:

        return "flu"


    return "unknown"


# --------------------------------------------------
# STEP 4:
# Test generalized behavior
# --------------------------------------------------
#
# Notice:
# the system now handles cases it has
# never seen directly before.
# --------------------------------------------------

print("\n=== GENERALIZED SYSTEM ===")


# Example with fever
print("\nPatient with fever:")

result_1 = generalize(True)

print("Diagnosis:", result_1)


# Example without fever
print("\nPatient without fever:")

result_2 = generalize(False)

print("Diagnosis:", result_2)


# --------------------------------------------------
# STEP 5:
# Compare the two systems
# --------------------------------------------------
#
# The memorized system:
# • performs perfectly on one exact example
# • fails immediately outside that example
#
# The generalized system:
# • uses a simpler structure
# • applies behavior across new situations
#
# Real machine learning systems constantly
# balance this tension:
#
# • too much memorization
#       weak generalization
#
# • too much simplification
#       missing important structure
# --------------------------------------------------


# --------------------------------------------------
# Think about this:
# --------------------------------------------------
#
# • The first system memorized one exact case
#
# • The second system applied a broader rule
#
# • Memorization may produce excellent
#   training performance
#
# • But real intelligence requires handling
#   unfamiliar situations successfully
#
# • Many AI failures only appear AFTER
#   deployment when new environments emerge
#
# This is one of the central tensions
# in machine learning:
#
# How can a system simplify reality enough
# to generalize...
#
# without losing the distinctions
# that actually matter?
#
# Key question:
# Is the system learning reusable structure...
# or storing isolated examples?
# --------------------------------------------------


print("\n=== DEMO COMPLETE ===")
```
