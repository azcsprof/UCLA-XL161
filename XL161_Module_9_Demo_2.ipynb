```python
# --------------------------------------------------
# Module 9 DEMO 2: Representation Shapes Learning
# --------------------------------------------------
# Different models organize the SAME data differently.
#
# Representation changes:
# • what distinctions become visible
# • what patterns can be separated
# • what the system learns
#
# Your job:
# 1. Compare how the system divides cases
# 2. Watch decisions form step-by-step
# 3. Notice how structure shapes behavior
# --------------------------------------------------


def diagnose(temperature, cough):

    # First split
    if temperature > 100:

        print("High temperature branch")

        # Second split
        if cough:

            print("Cough branch")
            return "flu"

        else:

            print("No cough branch")
            return "infection"

    else:

        print("Low temperature branch")
        return "cold"


# Test cases
cases = [
    (102, True),
    (102, False),
    (98, True)
]


# Run all examples
for temp, cough in cases:

    print("\nPatient:")
    print("Temperature:", temp)
    print("Cough:", cough)

    result = diagnose(temp, cough)

    print("Diagnosis:", result)


# --------------------------------------------------
# Think about this:
# --------------------------------------------------
# • The system is not learning randomly
#
# • It organizes the world through:
#       thresholds
#       branches
#       separations
#
# • Different representations would produce
#   different forms of learning
#
# Key question:
# Does data alone determine learning...
# or does representation shape what
# the system can notice?
# --------------------------------------------------
```
