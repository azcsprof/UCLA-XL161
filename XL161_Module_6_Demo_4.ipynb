```python
# -------------------------------------------
# Module 6 DEMO 4: When Facts Don't Connect
# -------------------------------------------
# The system can store facts as strings,
# but cannot represent relationships.
#
# Your job:
# 1. See what the system can retrieve
# 2. Notice what it cannot answer
# 3. Decide what is missing
# -------------------------------------------


facts = {"Alice_loves_Bob": True}


def query_alice_loves(facts):
    # Looks for a specific pattern in the string
    for key in facts:
        if key.startswith("Alice_loves_"):
            return key.split("_")[-1]
    return None


print("Stored facts:", facts)
print("Who does Alice love?", query_alice_loves(facts))


print("\n--- Add more data ---")

# Add another fact
facts["Bob_loves_Alice"] = True

print("Stored facts:", facts)
print("Who does Alice love?", query_alice_loves(facts))


# -------------------------------------------
# Think about this:
# -------------------------------------------
# • The system can retrieve exact patterns
# • Adding more facts does not improve reasoning
#
# Key question:
# Is the problem missing data...
# or missing structure?
# -------------------------------------------
```
