```python
# ==================================================
# Module 7 DEMO 4: When Matching Fails
# ==================================================
#
# Earlier demos showed that variables
# make rules reusable.
#
# But variables do not mean
# that any rule can apply anywhere.
#
# The structure still has to match.
#
# This process is called matching
# or, in more formal symbolic systems, unification.
#
# Matching checks whether a pattern and a fact
# can align through consistent variable assignment.
#
# Your job:
# 1. Watch a successful match
# 2. Watch a failed match
# 3. Notice where the conflict appears
# 4. Explain why structure controls inference
#
# Important:
# A rule can exist.
# A fact can exist.
# But inference still fails if the structures
# cannot align consistently.
# ==================================================


print("\n=== MODULE 7 DEMO 4 ===")
print("When Matching Fails")


# --------------------------------------------------
# STEP 1:
# Define a helper function
# --------------------------------------------------
#
# This function attempts to match:
# • a pattern with variables
# • a stored fact with concrete values
#
# Variables are represented as:
# x, y, z
#
# If a variable appears more than once,
# it must keep the same value every time.
# --------------------------------------------------

def match(pattern, fact):

    bindings = {}
    failed = False

    if len(pattern) != len(fact):
        return False, {}, "Length mismatch"

    for p, f in zip(pattern, fact):

        if p in ["x", "y", "z"]:

            if p in bindings:

                if bindings[p] != f:
                    failed = True
                    return False, bindings, "Conflict detected"

            else:
                bindings[p] = f

        else:

            if p != f:
                failed = True
                return False, bindings, "Predicate mismatch"

    return True, bindings, "Match succeeded"


# --------------------------------------------------
# STEP 2:
# Successful match
# --------------------------------------------------
#
# The variables x and y can be assigned
# without conflict.
# --------------------------------------------------

pattern = ("Loves", "x", "y")
fact = ("Loves", "alice", "bob")

print("\nSuccessful match attempt:")
print("Pattern:", pattern)
print("Fact:", fact)

success, bindings, message = match(pattern, fact)

print("Result:", message)
print("Bindings:", bindings)


# --------------------------------------------------
# STEP 3:
# Failed match
# --------------------------------------------------
#
# Now the pattern requires the same variable x
# to appear in both positions.
#
# That means both targets must be identical.
#
# But the fact contains alice and bob.
# --------------------------------------------------

pattern = ("Loves", "x", "x")
fact = ("Loves", "alice", "bob")

print("\nFailed match attempt:")
print("Pattern:", pattern)
print("Fact:", fact)

success, bindings, message = match(pattern, fact)

print("Result:", message)
print("Bindings before failure:", bindings)


# --------------------------------------------------
# STEP 4:
# Make the structures compatible
# --------------------------------------------------
#
# Now x can consistently become alice
# in both positions.
# --------------------------------------------------

pattern = ("Loves", "x", "x")
fact = ("Loves", "alice", "alice")

print("\nCompatible repeated variable:")
print("Pattern:", pattern)
print("Fact:", fact)

success, bindings, message = match(pattern, fact)

print("Result:", message)
print("Bindings:", bindings)


# --------------------------------------------------
# STEP 5:
# Length mismatch
# --------------------------------------------------
#
# Sometimes the structures fail
# before variables even matter.
# --------------------------------------------------

pattern = ("Parent", "x", "y")
fact = ("Parent", "alice")

print("\nLength mismatch example:")
print("Pattern:", pattern)
print("Fact:", fact)

success, bindings, message = match(pattern, fact)

print("Result:", message)


# --------------------------------------------------
# Think about this:
# --------------------------------------------------
#
# • Matching is not vague similarity
# • The structure must align
# • Variables must stay consistent
# • Small differences can stop reasoning completely
#
# This is why reusable rules do not apply automatically.
#
# Matching controls whether inference can continue.
#
# Key question:
# Why is a rule not enough by itself?
# --------------------------------------------------


print("\n=== DEMO COMPLETE ===")
```
