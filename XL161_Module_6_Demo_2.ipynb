```python
# -------------------------------------------
# Module 6 DEMO 2: When "Maybe" Becomes True/False
# -------------------------------------------
# This system must decide using True or False.
# It cannot represent uncertainty directly.
#
# Your job:
# 1. Compare how different values are treated
# 2. Notice what information is ignored
# 3. Decide whether the system understands uncertainty
# -------------------------------------------


def diagnose(fever):
    # The system treats anything "truthy" as True
    if fever:
        return "infection"
    else:
        return "no infection"


# Test cases include:
# • clear True/False
# • partial value (0.7)
# • absence of value (None)
cases = [True, False, 0.7, 0.0, None]


for value in cases:
    result = diagnose(value)
    print(f"fever = {value} -> {result}")


# -------------------------------------------
# Think about this:
# -------------------------------------------
# • 0.7 means "some evidence" to a human
#   → but the system treats it as fully True
#
# • None means "unknown"
#   → but the system treats it as False
#
# Key question:
# Did the system resolve uncertainty...
# or did it remove it?
# -------------------------------------------
```
