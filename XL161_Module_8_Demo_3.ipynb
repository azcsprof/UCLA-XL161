```python
# ============================================================
# MODULE 8 DEMO 3 — BELIEF UPDATES OVER TIME
# ============================================================

belief = {
    "flu": 0.5,
    "infection": 0.5
}

print("Starting beliefs:")
print(belief)

print()

updates = [
    ("fever", "flu", 0.2),
    ("chest pain", "infection", 0.3),
    ("negative_flu_test", "flu", -0.4)
]

for evidence, target, weight in updates:

    print("New evidence:")
    print(evidence)

    print()

    belief[target] += weight

    # Prevent negative probabilities.

    if belief[target] < 0:
        belief[target] = 0

    # Normalize beliefs.

    total = sum(belief.values())

    for key in belief:

        belief[key] /= total

    print("Updated beliefs:")
    print(belief)

    print()

# The system continuously
# reorganizes confidence as
# evidence arrives across time.

# Reasoning becomes gradual
# instead of binary.
```
