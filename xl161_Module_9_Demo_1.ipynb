```python
# --------------------------------------------------
# Module 9 DEMO 1: From Rules to Learned Patterns
# --------------------------------------------------
# Earlier AI systems depended on hand-written rules.
# Learning systems begin differently.
#
# Instead of specifying every rule directly,
# the system starts accumulating patterns
# across repeated examples.
#
# Your job:
# 1. Watch repeated examples accumulate
# 2. Observe how statistical regularities form
# 3. Notice that the system still does NOT
#    understand why the pattern exists
# --------------------------------------------------


# Example training cases
patients = [
    ("fever", "flu"),
    ("cough", "flu"),
    ("rash", "infection"),
    ("fever", "flu"),
    ("cough", "flu")
]


# Counters begin at zero
flu_count = 0
infection_count = 0


# The system processes examples one-by-one
for symptom, diagnosis in patients:

    print("\nNew example:", symptom, "->", diagnosis)

    # Pattern accumulation
    if diagnosis == "flu":
        flu_count += 1

    elif diagnosis == "infection":
        infection_count += 1

    # Current learned statistics
    print("Current flu count:", flu_count)
    print("Current infection count:", infection_count)


# Final pattern summary
print("\nFinal learned pattern:")
print("Flu:", flu_count)
print("Infection:", infection_count)


# --------------------------------------------------
# Think about this:
# --------------------------------------------------
# • The system was never explicitly told:
#       "fever means flu"
#
# • Instead, repeated examples gradually
#   strengthened statistical expectations
#
# • But notice:
#   the system still does not understand
#   medicine, illness, or causality
#
# Key question:
# Is the system learning meaning...
# or learning repeated statistical structure?
# --------------------------------------------------
```
