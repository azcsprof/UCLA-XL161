```python
# ==================================================
# Module 9 DEMO 1: From Rules to Learned Patterns
# ==================================================
#
# Earlier AI systems depended heavily on rules.
# Humans had to explicitly specify:
# • categories
# • relationships
# • conditions
# • expected outcomes
#
# Learning systems approach the problem differently.
# Instead of manually writing every rule,
# the system begins accumulating patterns
# from repeated examples.
#
# In this demo, you will watch:
# • examples accumulate over time
# • statistical regularities begin forming
# • predictions emerge from repetition
#
# Your job:
# 1. Watch repeated examples strengthen patterns
# 2. Observe how counts change over time
# 3. Notice that the system tracks frequency,
#    not meaning or causality
#
# Important:
# The system does NOT understand illness.
# It only accumulates repeated statistical structure.
# ==================================================


print("\n=== MODULE 9 DEMO 1 ===")
print("From Rules to Learned Patterns")


# --------------------------------------------------
# STEP 1:
# Create example training data
# --------------------------------------------------
#
# Each example contains:
# • a symptom
# • an associated diagnosis
#
# This acts like simplified "experience"
# for the learning system.
# --------------------------------------------------

patients = [
    ("fever", "flu"),
    ("cough", "flu"),
    ("rash", "infection"),
    ("fever", "flu"),
    ("cough", "flu")
]


# --------------------------------------------------
# STEP 2:
# Create counters for learned patterns
# --------------------------------------------------
#
# The system will gradually accumulate
# statistical regularities by counting
# repeated examples.
#
# Initially:
# • the system knows nothing
# • all counts begin at zero
# --------------------------------------------------

flu_count = 0
infection_count = 0


print("\nProcessing training examples...")


# --------------------------------------------------
# STEP 3:
# Process examples one-by-one
# --------------------------------------------------
#
# The loop simulates the system observing
# repeated experiences across time.
#
# After each example:
# • the internal counts change
# • the learned statistics update
# --------------------------------------------------

for symptom, diagnosis in patients:

    print("\nObserved example:")
    print("Symptom:", symptom)
    print("Diagnosis:", diagnosis)


    # ----------------------------------------------
    # Update internal statistical patterns
    # ----------------------------------------------
    #
    # The system strengthens whichever diagnosis
    # appears repeatedly in the training data.
    #
    # This is a simplified version of
    # statistical learning.
    # ----------------------------------------------

    if diagnosis == "flu":
        flu_count += 1

    elif diagnosis == "infection":
        infection_count += 1


    # ----------------------------------------------
    # Show current internal state
    # ----------------------------------------------
    #
    # These values represent the model's
    # gradually accumulated expectations.
    # ----------------------------------------------

    print("\nCurrent learned statistics:")
    print("Flu count:", flu_count)
    print("Infection count:", infection_count)


# --------------------------------------------------
# STEP 4:
# Generate a prediction from accumulated patterns
# --------------------------------------------------
#
# The system now compares the learned counts.
#
# Whichever diagnosis appeared more often
# becomes the strongest statistical expectation.
# --------------------------------------------------

if flu_count > infection_count:
    prediction = "flu"

else:
    prediction = "infection"


# --------------------------------------------------
# STEP 5:
# Show final learned result
# --------------------------------------------------

print("\n=== FINAL LEARNED PATTERNS ===")
print("Total flu examples:", flu_count)
print("Total infection examples:", infection_count)

print("\nMost statistically likely diagnosis:")
print(prediction)


# --------------------------------------------------
# Think about this:
# --------------------------------------------------
#
# • The system was never explicitly told:
#       "fever causes flu"
#
# • It simply accumulated repeated examples
#
# • The learned behavior emerged from
#   statistical repetition across data
#
# • But notice:
#   the system still does NOT understand:
#       illness
#       biology
#       causality
#
# • It only tracks repeated structure
#
# This is one of the central transitions
# in machine learning:
#
# from:
#     hand-written knowledge
#
# to:
#     learned statistical regularities
#
# Key question:
# Is the system learning meaning...
# or learning repeated statistical patterns?
# --------------------------------------------------


print("\n=== DEMO COMPLETE ===")
```
