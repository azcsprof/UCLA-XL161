```python
# ============================================================
# MODULE 8 DEMO 1 — WHY ELIMINATION BREAKS
# ============================================================

symptoms = ["fever", "cough"]

possible_explanations = []

print("Symptoms:")
print(symptoms)

print()

# Fever + cough may indicate flu.

if "fever" in symptoms and "cough" in symptoms:

    possible_explanations.append("flu")

# Fever alone may still indicate infection.

if "fever" in symptoms:

    possible_explanations.append("infection")

print("Possible explanations:")
print(possible_explanations)

print()

# Earlier symbolic systems expected
# one clean elimination path.

# But now several explanations
# survive simultaneously.

# The environment remains ambiguous.
```
