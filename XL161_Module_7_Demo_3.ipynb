```python
# ==================================================
# Module 7 DEMO 3: Rules That Travel
# ==================================================
#
# Demo 2 showed that structured relationships
# make flexible querying possible.
#
# But symbolic systems can do more than query.
#
# They can also apply reusable rules.
#
# A reusable rule contains a variable.
#
# Example:
# Human(x) -> Mortal(x)
#
# This rule is not about one specific person.
# It can apply to any object that matches Human(x).
#
# Your job:
# 1. Watch one rule apply to multiple facts
# 2. Notice what the variable makes possible
# 3. Explain why this is reuse, not machine learning
# 4. Identify what new conclusions are derived
#
# Important:
# The system is not discovering a rule from data.
# The rule is already written.
#
# The system is reusing the rule wherever
# compatible structure appears.
# ==================================================


print("\n=== MODULE 7 DEMO 3 ===")
print("Rules That Travel")


# --------------------------------------------------
# STEP 1:
# Store known facts
# --------------------------------------------------
#
# Each fact has:
# • a predicate
# • an object
#
# These facts say who is human.
# --------------------------------------------------

facts = [
    ("Human", "alice"),
    ("Human", "bob"),
    ("Human", "dana")
]


print("\nKnown facts:")

for fact in facts:
    print(fact)


# --------------------------------------------------
# STEP 2:
# Define the reusable rule
# --------------------------------------------------
#
# Informally:
#
# If Human(x), then Mortal(x)
#
# In this simple demo, we represent the rule
# through code rather than a full logic engine.
# --------------------------------------------------

print("\nRule:")
print("Human(x) -> Mortal(x)")


# --------------------------------------------------
# STEP 3:
# Apply the rule across matching facts
# --------------------------------------------------
#
# The system checks each fact.
#
# If the fact has the predicate "Human",
# the system derives a new fact:
#
# Mortal(same object)
# --------------------------------------------------

derived = []

for predicate, obj in facts:

    if predicate == "Human":

        new_fact = ("Mortal", obj)

        derived.append(new_fact)


print("\nDerived conclusions:")

for fact in derived:
    print(fact)


# --------------------------------------------------
# STEP 4:
# Add a new fact
# --------------------------------------------------
#
# Now a new human appears.
#
# We do not change the rule.
# The same rule should apply again.
# --------------------------------------------------

facts.append(("Human", "samara"))

print("\nAfter adding a new fact:")
print(("Human", "samara"))


derived = []

for predicate, obj in facts:

    if predicate == "Human":

        derived.append(("Mortal", obj))


print("\nDerived conclusions after update:")

for fact in derived:
    print(fact)


# --------------------------------------------------
# STEP 5:
# Break the rule condition
# --------------------------------------------------
#
# Now remove the protection.
#
# Instead of applying the rule only to humans,
# force it to apply to everything.
#
# This shows why matching matters.
# --------------------------------------------------

facts.append(("Robot", "unit7"))

broken_derived = []

for predicate, obj in facts:

    if True:

        broken_derived.append(("Mortal", obj))


print("\nAfter breaking the condition:")
print("Derived conclusions:")

for fact in broken_derived:
    print(fact)


# --------------------------------------------------
# Think about this:
# --------------------------------------------------
#
# • One rule applied to many objects
# • A new fact triggered the same rule automatically
# • The rule did not need to be rewritten
# • But removing the condition made the rule too broad
#
# This is symbolic reuse.
#
# The system is not learning statistically.
#
# It is applying a reusable pattern
# across compatible structures.
#
# Key question:
# What protects a reusable rule from applying
# to the wrong things?
# --------------------------------------------------


print("\n=== DEMO COMPLETE ===")
```
