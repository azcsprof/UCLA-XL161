```python
# -------------------------------------------
# Moudle 6 DEMO 1: Elimination vs Construction
# -------------------------------------------
# This demo compares two systems that produce
# the SAME final result, but in very different ways.
#
# Your job:
# 1. Watch WHEN invalid states are handled
# 2. Compare how much unnecessary work is done
# 3. Decide which system is actually "reasoning"
# -------------------------------------------


def naive():
    # The system will try ALL combinations first,
    # then check if they are valid.

    colors = ["Red", "Green"]
    states = []
    built = 0

    for A in colors:
        for B in colors:
            built += 1  # counts every state that is constructed

            # ⚠️ Constraint applied AFTER building the state
            # The system already created (A, B), even if invalid
            if A != B:
                states.append((A, B))

    print("Naive valid states:", states)
    print("States built:", built)


def elimination():
    # This system checks constraints BEFORE completing the state.
    # Invalid combinations are never fully constructed.

    colors = ["Red", "Green"]
    states = []
    built = 0

    for A in colors:
        for B in colors:

            # ⚠️ Constraint applied BEFORE building the state
            # If invalid, skip immediately (state never exists)
            if B == A:
                continue

            built += 1  # only counts states that pass constraint
            states.append((A, B))

    print("\nWith elimination valid states:", states)
    print("States built:", built)


# -------------------------------------------
# Run both systems
# -------------------------------------------
naive()
elimination()


# -------------------------------------------
# Think about this:
# -------------------------------------------
# • Both systems return the same valid states.
# • But which one does more unnecessary work?
# • Which system "knows" earlier that a state is invalid?
#
# Key question:
# Is the improvement happening at the level of
# checking... or at the level of construction?
# -------------------------------------------
```
