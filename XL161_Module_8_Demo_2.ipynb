```python
# ============================================================
# MODULE 8 DEMO 2 — WEIGHTED BELIEF
# ============================================================

belief = {
    "flu": 0.6,
    "infection": 0.3,
    "allergy": 0.1
}

print("Initial beliefs:")
print(belief)

print()

# The system compares explanations.

best_guess = max(
    belief,
    key=belief.get
)

print("Strongest explanation:")
print(best_guess)

print()

# Confidence must still behave
# like a probability distribution.

total = sum(belief.values())

for key in belief:

    belief[key] /= total

print("Normalized beliefs:")
print(belief)

print()

# Stronger explanations receive
# more confidence.

# But weaker explanations
# still remain alive internally.
```
