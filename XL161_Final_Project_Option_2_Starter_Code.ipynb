```python
# ============================================================
# COM SCI XL161 FINAL PROJECT — OPTION 2
# Intelligent Diagnostic Reasoning System
# ============================================================
#
# Core Idea:
#
# This project models a simple AI diagnostic assistant.
#
# The system receives symptoms from a user and attempts to:
#
#   • reason using logical rules
#   • represent relationships between symptoms and diseases
#   • handle uncertainty using probabilities
#   • learn from diagnostic outcomes
#
# The important question is NOT:
#
#     "Did the system guess correctly?"
#
# The important question is:
#
#     "How did the system reason?"
#
# ============================================================
#
# MODULE CONNECTIONS
#
# Module 1:
#   Agents, states, goals, representation
#
# Module 2:
#   Search through possible diagnoses
#
# Module 3:
#   Heuristic prioritization of likely diagnoses
#
# Module 4:
#   Constraint-based elimination
#
# Module 6:
#   Rule-based inference
#
# Module 7:
#   Structured relationships and predicates
#
# Module 8:
#   Probabilistic reasoning / uncertainty
#
# Module 9:
#   Learning from outcomes
#
# ============================================================


# ============================================================
# 1. POSSIBLE DIAGNOSES
# ============================================================
#
# The system must choose between several possible explanations.
#
# Each diagnosis represents a possible "goal state"
# the system may conclude.
#
# ============================================================

diagnoses = [
    "Flu",
    "Cold",
    "Allergy",
    "Pneumonia"
]


# ============================================================
# 2. USER SYMPTOMS
# ============================================================
#
# The system receives observations from the environment.
#
# In this project, the observations are symptoms.
#
# These symptoms become the evidence used for reasoning.
#
# ============================================================

patient_symptoms = {
    "fever": True,
    "cough": True,
    "fatigue": True,
    "shortness_of_breath": False,
    "sneezing": False
}


# ============================================================
# 3. RULE-BASED KNOWLEDGE
# ============================================================
#
# Module 6 connection:
#
# These are symbolic IF/THEN rules.
#
# Example:
#
# IF fever AND cough
# THEN Flu becomes more likely.
#
# The system is not learning these rules.
# Humans encoded them directly.
#
# ============================================================

rules = {
    "Flu": ["fever", "cough", "fatigue"],
    "Cold": ["cough"],
    "Allergy": ["sneezing"],
    "Pneumonia": ["fever", "cough", "shortness_of_breath"]
}


# ============================================================
# 4. PROBABILISTIC BELIEFS
# ============================================================
#
# Module 8 connection:
#
# Logical rules alone create rigid true/false reasoning.
#
# Real environments contain uncertainty.
#
# These probabilities represent:
#
#     "How likely is this diagnosis BEFORE reasoning?"
#
# These are prior probabilities.
#
# ============================================================

diagnosis_probability = {
    "Flu": 0.40,
    "Cold": 0.30,
    "Allergy": 0.20,
    "Pneumonia": 0.10
}


# ============================================================
# 5. LEARNED ADJUSTMENTS
# ============================================================
#
# Module 9 connection:
#
# The system can learn from previous outcomes.
#
# If a diagnosis repeatedly succeeds or fails,
# its score can be adjusted over time.
#
# This is a simplified learning mechanism.
#
# ============================================================

learned_adjustments = {
    "Flu": 0,
    "Cold": 0,
    "Allergy": 0,
    "Pneumonia": 0
}


# ============================================================
# 6. REPRESENTATION FUNCTION
# ============================================================
#
# Module 7 connection:
#
# This function creates structured relationships between:
#
#     patient → symptom
#
# Instead of storing isolated facts,
# the system organizes information relationally.
#
# Example:
#
#     has_symptom(patient, fever)
#
# This moves beyond flat propositional statements.
#
# ============================================================

def represent_patient_knowledge():
    print("\n=== STRUCTURED REPRESENTATION ===")

    for symptom, present in patient_symptoms.items():

        if present:
            print(f"has_symptom(patient, {symptom})")


# ============================================================
# 7. RULE-BASED REASONING
# ============================================================
#
# Module 6 connection:
#
# This function applies symbolic rules.
#
# The system checks:
#
#     Which symptoms match each diagnosis?
#
# The more matching rules,
# the stronger the symbolic support.
#
# ============================================================

def rule_based_reasoning():
    print("\n=== RULE-BASED REASONING ===")

    rule_scores = {}

    for diagnosis in diagnoses:

        matching_symptoms = 0

        print(f"\nEvaluating diagnosis: {diagnosis}")

        for symptom in rules[diagnosis]:

            if patient_symptoms.get(symptom, False):
                matching_symptoms += 1

                print(f"  Rule matched: {symptom}")

        rule_scores[diagnosis] = matching_symptoms

        print(f"  Total matching rules: {matching_symptoms}")

    return rule_scores


# ============================================================
# 8. CONSTRAINT-BASED ELIMINATION
# ============================================================
#
# Module 4 connection:
#
# Some diagnoses become impossible if critical symptoms
# are missing.
#
# Example:
#
# Pneumonia requires shortness_of_breath.
#
# If that symptom is absent,
# we eliminate Pneumonia entirely.
#
# This is elimination through constraints.
#
# ============================================================

def apply_constraints(possible_diagnoses):
    print("\n=== CONSTRAINT CHECKING ===")

    remaining = []

    for diagnosis in possible_diagnoses:

        # Pneumonia constraint
        if diagnosis == "Pneumonia":

            if not patient_symptoms["shortness_of_breath"]:
                print("Eliminating Pneumonia:")
                print("  Missing required symptom: shortness_of_breath")

                continue

        remaining.append(diagnosis)

    return remaining


# ============================================================
# 9. HEURISTIC PRIORITIZATION
# ============================================================
#
# Module 3 connection:
#
# Search systems often prioritize more promising possibilities.
#
# Here, the system prioritizes diagnoses with:
#
#   • stronger rule support
#   • higher probabilities
#   • positive learned adjustments
#
# This acts like a heuristic scoring function.
#
# ============================================================

def compute_diagnosis_scores(rule_scores):
    print("\n=== DIAGNOSIS SCORING ===")

    diagnosis_scores = {}

    for diagnosis in diagnoses:

        probability = diagnosis_probability[diagnosis]
        learned_bonus = learned_adjustments[diagnosis]
        symbolic_support = rule_scores[diagnosis]

        final_score = (
            probability * 5
            + symbolic_support * 2
            + learned_bonus
        )

        diagnosis_scores[diagnosis] = final_score

        print(f"\nDiagnosis: {diagnosis}")
        print(f"  Prior probability: {probability}")
        print(f"  Symbolic support: {symbolic_support}")
        print(f"  Learned adjustment: {learned_bonus}")
        print(f"  Final score: {final_score:.2f}")

    return diagnosis_scores


# ============================================================
# 10. SEARCH THROUGH POSSIBLE DIAGNOSES
# ============================================================
#
# Module 2 connection:
#
# The system explores multiple candidate diagnoses.
#
# Rather than immediately committing,
# it evaluates alternatives and selects the best option.
#
# ============================================================

def choose_best_diagnosis(diagnosis_scores, valid_diagnoses):
    print("\n=== SEARCHING FOR BEST DIAGNOSIS ===")

    best_diagnosis = None
    best_score = -1

    for diagnosis in valid_diagnoses:

        score = diagnosis_scores[diagnosis]

        print(f"Considering: {diagnosis}")
        print(f"  Score: {score:.2f}")

        if score > best_score:
            best_score = score
            best_diagnosis = diagnosis

    return best_diagnosis


# ============================================================
# 11. LEARNING FROM OUTCOMES
# ============================================================
#
# Module 9 connection:
#
# After the system makes a diagnosis,
# we simulate feedback.
#
# If the diagnosis was correct:
#     increase future preference
#
# If incorrect:
#     decrease future preference
#
# This creates adaptation over time.
#
# ============================================================

def update_learning(correct_diagnosis, predicted_diagnosis):
    print("\n=== LEARNING UPDATE ===")

    if correct_diagnosis == predicted_diagnosis:

        learned_adjustments[predicted_diagnosis] += 1

        print("Prediction was correct.")
        print(f"Increasing learned preference for {predicted_diagnosis}")

    else:
        learned_adjustments[predicted_diagnosis] -= 1

        print("Prediction was incorrect.")
        print(f"Reducing learned preference for {predicted_diagnosis}")

    print("\nUpdated learned adjustments:")

    for diagnosis, value in learned_adjustments.items():
        print(f"  {diagnosis}: {value}")


# ============================================================
# 12. FULL SYSTEM EXECUTION
# ============================================================
#
# This runs the entire reasoning process:
#
#   1. Represent knowledge
#   2. Apply rules
#   3. Apply constraints
#   4. Score diagnoses
#   5. Search for best diagnosis
#   6. Learn from feedback
#
# ============================================================

def run_diagnostic_system(correct_answer):
    print("\n================================================")
    print("INTELLIGENT DIAGNOSTIC SYSTEM")
    print("================================================")

    represent_patient_knowledge()

    rule_scores = rule_based_reasoning()

    valid_diagnoses = apply_constraints(diagnoses)

    diagnosis_scores = compute_diagnosis_scores(rule_scores)

    predicted_diagnosis = choose_best_diagnosis(
        diagnosis_scores,
        valid_diagnoses
    )

    print("\n=== FINAL DIAGNOSIS ===")
    print(f"Predicted diagnosis: {predicted_diagnosis}")

    update_learning(correct_answer, predicted_diagnosis)


# ============================================================
# 13. RUN THE SYSTEM
# ============================================================
#
# Run the system multiple times and observe:
#
#   • how rules influence reasoning
#   • how constraints eliminate possibilities
#   • how probabilities shape decisions
#   • how learning changes future behavior
#
# ============================================================

run_diagnostic_system(correct_answer="Flu")

print("\n\nSECOND RUN AFTER LEARNING:\n")

run_diagnostic_system(correct_answer="Flu")

# STUDENT EXTENSIONS
#
# 1. Add a new disease.
#
# 2. Add conflicting symptoms.
#
# 3. Add noisy sensor input:
#       symptom may be uncertain.
#
# 4. Modify the rule system.
#
# 5. Change probabilities.
#
# 6. Add weighted symptom importance.
#
# 7. Break the system intentionally:
#       bad probabilities
#       incorrect rules
#       missing constraints
#
# 8. Improve the system:
#       stronger rules
#       better learning
#       improved uncertainty handling
```
